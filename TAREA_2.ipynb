# Imports

```python
import matplotlib.pyplot as plt
import numpy as np
import math
import pandas as pd
from pandas import DataFrame, Series
```

# Función

```python
def f(tD_xD_2):
    return (2 
            * math.sqrt(tD_xD_2/np.pi) 
            * math.exp(-(1 / (4 * tD_xD_2)))
           - math.erfc(1/ (2 * math.sqrt(tD_xD_2)))
           )
```

# Gráfica

```python
x = np.logspace(-2,3, 50)
y = list(map(f,x))

plt.plot(x,y)

plt.xscale('log')
plt.yscale('log')

plt.xlabel("$t_D/x_D^2$")
plt.ylabel(r"$p_D/x_D$")

plt.title("Curva tipo, flujo lineal")

plt.tight_layout()

plt.grid(True)
```

```python
data = pd.DataFrame()
data[r"$t_D/x_D^2$"] = x
data[r"$p_D/x_D$"] = y
```
